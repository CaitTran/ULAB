## 1 Functions
**Orbital Periods and Kepler's Third Law** \
Kepler's Third Law states that the square of the orbital period $T$ (in years) of a planet is proportional to the cube of the semi-major axis $a$ (in astronomical units, AU):

$$
T^2 = a^3
$$

Write a **function** that calculates the orbital period $T$ given the semi-major axis $a$. (So, you can assume that $T = \sqrt{a^3}$.)

```python
def orbital_periods(a):
    #Gives orbital period given semi-major axis (Kepler's Third Law)
    period_T = (a**3)**0.5
    return period_T
```

Call your function with the following values: 
* `planet_a` $ = 0.5\,\text{AU}$
* `planet_b` $ = 1.0\,\text{AU}$
* `planet_c` $ = 1.5\,\text{AU}$

```python
planet_a = 0.5
planet_b = 1.0
planet_c = 1.5

orbital_periods(planet_a)

orbital_periods(planet_b)

orbital_periods(planet_c)
```

Call both of your functions in the cell below with ($a$ = 5). Time how long it takes each of your functions to run. **DO NOT REWRITE THEM!** 

```python
#time.perf_counter() creates a ticker that counts time at a certain point, calculating the difference between the two points gives the elapsed time
start_time_1 = time.perf_counter()

orbital_periods(a=5)

end_time_1 = time.perf_counter()

elapsed_time_1 = end_time_1 - start_time_1

start_time_2 = time.perf_counter()

orbital_periods_2(a=5)

end_time_2 = time.perf_counter()

elapsed_time_2 = end_time_2 - start_time_2

print(elapsed_time_1, elapsed_time_2)
```

## 2 Indexing + Conditionals

You are provided with a list of exoplanet radii (in Earth radii) and their equilibrium temperatures (in Kelvin):

`radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]` \
`temperatures = [300, 400, 350, 280, 500, 450]`

Write a loop to filter and print the radii of planets with temperatures below 350 K.

```python
def find_planets(temperatures):
    #Prints planets with temperatures below 350 K
    filtered_tempplanets = []
    for i in range(len(temperatures)):
        #for the index i within the range of the length of the list of temperatures
        if temperatures[i] < 350:
            #if the value corresponding to the index is less than 350
            filtered_tempplanets.append(temperatures[i])
            #append the value to the new list
    return filtered_tempplanets

temperatures = [300, 400, 350, 280, 500, 450]
find_planets(temperatures)
```

Use list slicing to extract and print the radii of the first three planets.

```python
def first_three(radii):
    #Gives the first three planets' radii from the list of radii
    filtered_radiiplanets = []
    for i in range(len(radii)):
        #for the index within the range of the length of the radii list
        if i < 3:
            #if the index is less than three (starting from 0) 
            filtered_radiiplanets.append(radii[i])
            #append the value corresponding to the index to the new list
    return filtered_radiiplanets

radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]
first_three(radii)
```

Turn the radii and temperatures data into a dictionary. 

```python
radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]
temperatures =  [300, 400, 350, 280, 500, 450]

#takes the 2 lists and pairs each corresponding index together, dictionary converts the tuples into a dictionary
radii_and_temps_dictionary = dict(zip(radii, temperatures))
print(radii_and_temps_dictionary)
```

Give each of the planets in your dictionary a name. 

```python
planet_names = ["PlanetA", "PlanetB", "PlanetC", "PlanetD", "PlanetE", "PlanetF"]

planet_dictionary = {}
#initialize the dictionary
for planet_name, radius, temp in zip(planet_names, radii, temperatures):
    #for each corresponding index within the paired set of 3 lists
    planet_dictionary[planet_name] = {"radius": radius, "temperature": temp}
    #add to the dictionary the planet_name index value as the key, the radius and temperature as attributes with their own keys & values
    
print(planet_dictionary)
```

Create a new dictionary of planets with radii between 1.0 and 2.0 Earth radii (inclusive).

```python
filtered_planets = {}
#initialize dictionary
for key, data in planet_dictionary.items():
    #for the key and data within the previous dictionary
    if 1.0 <= data["radius"] <= 2.0:
        #if the data for radius has a value less than 2 and greater than 1
        filtered_planets[key] = data
        #append the data to the new dictionary

print(filtered_planets)
```

## 4 Plotting
Create a scatter plot of the following semi-major axes (x-axis) vs. orbital periods (y-axis). Label the axes, add a legend and add a title: "Orbital Periods of Exoplanets."

`semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]`

```python
import matplotlib.pyplot as plt
```

```python
semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]

y_data = []
for axis in semi_major_axes:
    orbital_period = orbital_periods_2(axis)
    y_data.append(orbital_period)

plt.plot
plt.scatter(semi_major_axes, y_data)
plt.xlabel("Semi-Major Axis")
plt.ylabel("Orbital Period")
plt.title("Orbital Periods of Exoplanets")
plt.legend("Planets")
plt.show()
```

Create a bar plot of planet radii vs. their equilibrium temperatures. Label the axes, add a legend and add a title: "Equilibrium Temperatures vs. Radii."

```python
radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]
temperatures = [300, 400, 350, 280, 500, 450]

plt.plot
plt.bar(radii, temperatures)
plt.title("Equilibrium Temperatures vs. Radii")
plt.xlabel("Radius")
plt.ylabel("Equilibrium Temperature")
plt.legend("Planets")
plt.show()
```

Put the scatter plot and bar plot on the same figure (horizonatally). 

```python
fig, ax = plt.subplots(1, 2)
ax[0].scatter(semi_major_axes, y_data)
ax[0].set_xlabel("Semi-Major Axis")
ax[0].set_ylabel("Orbital Period")
ax[0].set_title("Orbital Periods of Exoplanets")
ax[0].legend("Planets")

ax[1].bar(radii, temperatures)
ax[1].set_title("Equilibrium Temperatures vs. Radii")
ax[1].set_xlabel("Radius")
ax[1].set_ylabel("Equilibrium Temperature")
ax[1].legend("Planets")

plt.subplots_adjust(wspace=0.5)

plt.show()
```
